```python
##以最小文件夹为单位显示所有图片
import os
from PIL import Image
import matplotlib.pyplot as plt
import math


def find_leaf_folders(root_dir, exts=(".jpg", ".jpeg", ".png")):
    """
    找到所有最小文件夹（叶子目录）：
    该目录下没有子文件夹，并且包含指定后缀的图片文件
    """
    leaf_folders = []

    for current_root, dirs, files in os.walk(root_dir):
        has_subdirs = len(dirs) > 0
        has_images = any(f.lower().endswith(exts) for f in files)

        if (not has_subdirs) and has_images:
            leaf_folders.append(current_root)

    return sorted(leaf_folders)


def show_images_in_folder(folder_path, exts=(".jpg", ".jpeg", ".png"), cols=4, figsize_per_img=3):
    """
    显示某个文件夹下所有图片，并打印该文件夹路径
    """
    image_files = sorted([
        f for f in os.listdir(folder_path)
        if f.lower().endswith(exts)
    ])

    if len(image_files) == 0:
        return

    print("=" * 120)
    print("最小文件夹路径：")
    print(folder_path)
    print(f"图片数量: {len(image_files)}")

    n = len(image_files)
    rows = math.ceil(n / cols)

    plt.figure(figsize=(cols * figsize_per_img, rows * figsize_per_img))

    for i, file_name in enumerate(image_files, 1):
        img_path = os.path.join(folder_path, file_name)
        img = Image.open(img_path)

        plt.subplot(rows, cols, i)
        plt.imshow(img)
        plt.title(file_name, fontsize=9)
        plt.axis("off")

    plt.tight_layout()
    plt.show()



##显示图片
root_dir = r"G:\wxy_codes\Fetal_hydronephrosis\第二批数据及处理过程\batch2data_3\raw\case_0014"   # 改成你的jpg根目录

leaf_folders = find_leaf_folders(root_dir)

print(f"共找到 {len(leaf_folders)} 个最小文件夹\n")

for folder in leaf_folders:
    show_images_in_folder(folder, cols=4, figsize_per_img=3)
```
