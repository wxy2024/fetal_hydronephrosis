```python
##打印出所有检查文件夹的下一级文件夹的路径，查看有没有除了0以外其他的文件名
from pathlib import Path

root_dir = Path(r"G:\wxy_codes\Fetal_hydronephrosis\1第一批数据\batch1data_1\raw")

result_dirs = []

for exam_dir in root_dir.rglob("*"):
    # 检查文件夹：文件夹名以数字开头，例如 202502140541
    if exam_dir.is_dir() and exam_dir.name[:1].isdigit():
        
        # 只收集检查文件夹下一级的文件夹
        for child in exam_dir.iterdir():
            if child.is_dir():
                result_dirs.append(child)

print(f"共找到 {len(result_dirs)} 个下一级子文件夹路径：\n")

for p in sorted(result_dirs):
    print(p)
```

```python
#把名字是0或者0_JPG的文件名字改成images

from pathlib import Path

root_dir = Path(r"G:\wxy_codes\Fetal_hydronephrosis\1第一批数据\batch1data_1\raw")

for case_dir in root_dir.iterdir():
    if not case_dir.is_dir():
        continue

    for stage_dir in case_dir.iterdir():
        if not stage_dir.is_dir():
            continue

        for visit_dir in stage_dir.iterdir():
            if not visit_dir.is_dir():
                continue

            folder_0 = visit_dir / "0"
            folder_0_jpg = visit_dir / "0_JPG"
            folder_images = visit_dir / "images"

            has_0 = folder_0.exists() and folder_0.is_dir()
            has_0_jpg = folder_0_jpg.exists() and folder_0_jpg.is_dir()
            has_images = folder_images.exists()

            # 同时存在 0 和 0_JPG，跳过
            if has_0 and has_0_jpg:
                print(f"[跳过] 同时存在 '0' 和 '0_JPG': {visit_dir}")
                continue

            # 目标 images 已存在，也跳过
            if has_images:
                print(f"[跳过] 已存在 'images' 文件夹: {folder_images}")
                continue

            # 只有 0
            if has_0:
                try:
                    folder_0.rename(folder_images)
                    print(f"[已重命名] {folder_0}  ->  {folder_images}")
                except Exception as e:
                    print(f"[失败] {folder_0} 重命名失败: {e}")

            # 只有 0_JPG
            elif has_0_jpg:
                try:
                    folder_0_jpg.rename(folder_images)
                    print(f"[已重命名] {folder_0_jpg}  ->  {folder_images}")
                except Exception as e:
                    print(f"[失败] {folder_0_jpg} 重命名失败: {e}")

            # 两者都不存在
            else:
                pass

print("处理完成")
```
