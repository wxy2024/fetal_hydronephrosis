```python
##展示所有dat文件的路径
import os

# ======================
# 只需要改这里！填入你要扫描的顶层文件夹路径
# ======================
root_dir = r"G:\wxy_codes\Fetal_hydronephrosis\1第一批数据\前期数据20260320\图像"

# 开始查找所有 .dat 文件
dat_files = []

for dirpath, dirnames, filenames in os.walk(root_dir):
    for filename in filenames:
        # 只找后缀是 .dat 的文件
        if filename.lower().endswith(".dat"):
            full_path = os.path.join(dirpath, filename)
            dat_files.append(full_path)
            print(full_path)

# 最后统计总数
print("\n✅ 一共找到", len(dat_files), "个 .dat 文件")
```

```python
##删除找到的dat文件，注意：要先执行上面的代码找到dat文件路径，然后再执行下面的代码删除这些文件
for fp in dat_files:
    try:
        os.remove(fp)
        print(f"已删除：{fp}")
    except Exception as e:
        print(f"删除失败 {fp}：{e}")

print("\n删除操作执行完毕")
```

```python
##删除Viewer文件夹，删除DICOMDIR文件，删除Viewer.exe文件
from pathlib import Path
import shutil

# 根目录
root = Path(r"G:\wxy_codes\Fetal_hydronephrosis\1第一批数据\前期数据20260320\图像")

# 是否只预览，不真正删除
dry_run = False   # 确认无误后改成 False

# 需要删除的文件夹名和文件名
target_dirs = {"Viewer"}
target_files = {"DICOMDIR", "Viewer.exe"}

deleted_dirs = []
deleted_files = []

for current_dir, dirs, files in os.walk(root):
    current_path = Path(current_dir)

    # 删除目标文件夹
    for d in dirs[:]:
        if d in target_dirs:
            dir_path = current_path / d
            print(f"[删除文件夹] {dir_path}")

            if not dry_run:
                shutil.rmtree(dir_path)

            deleted_dirs.append(dir_path)

            # 防止 os.walk 继续进入这个已删除的文件夹
            dirs.remove(d)

    # 删除目标文件
    for f in files:
        if f in target_files:
            file_path = current_path / f
            print(f"[删除文件] {file_path}")

            if not dry_run:
                file_path.unlink()

            deleted_files.append(file_path)

print("\n处理完成")
print(f"匹配到文件夹数量：{len(deleted_dirs)}")
print(f"匹配到文件数量：{len(deleted_files)}")

if dry_run:
    print("\n当前是预览模式，没有真正删除。确认无误后，把 dry_run = False 再运行一次。")
```
