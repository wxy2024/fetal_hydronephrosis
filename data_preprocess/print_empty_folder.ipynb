找到raw文件夹下所有的空文件夹打印出来路径

```python
from pathlib import Path

root_dir = Path(r"G:\wxy_codes\Fetal_hydronephrosis\第二批数据及处理过程\batch2data_1\raw")

ignore_files = {".ds_store", "thumbs.db"}

empty_dirs = []

for dir_path in root_dir.rglob("*"):
    if dir_path.is_dir():
        valid_items = [
            p for p in dir_path.iterdir()
            if p.name.lower() not in ignore_files
        ]
        if len(valid_items) == 0:
            empty_dirs.append(dir_path)

print(f"共找到 {len(empty_dirs)} 个空文件夹（忽略系统文件）：\n")

for d in empty_dirs:
    print(d)
```
