```python
##把root_dir的每一个子文件夹中都创建三个子文件夹和一个case_meta.json文件
import os
import json

# 根目录
##例：G:\wxy_codes\Fetal_hydronephrosis\fetal_hydro_dataset\raw
root_dir = r"G:\wxy_codes\Fetal_hydronephrosis\第二批数据及处理过程\batch2data_1\raw"

# 需要创建的子文件夹名称
subfolders = ["1prenatal", "2postnatal", "3postoperative"]

# 默认的 case_meta.json 内容
default_case_meta = {
    "case_id": "",
    "left_label": 0,
    "right_label": 0
}

# 检查根目录是否存在
if not os.path.exists(root_dir):
    raise FileNotFoundError(f"根目录不存在: {root_dir}")

# 遍历 raw 下的每个一级子文件夹
for name in os.listdir(root_dir):
    case_path = os.path.join(root_dir, name)
    
    # 只处理文件夹
    if not os.path.isdir(case_path):
        continue

    print(f"\n处理: {case_path}")

    # 1. 创建三个子文件夹
    for folder_name in subfolders:
        folder_path = os.path.join(case_path, folder_name)
        if not os.path.exists(folder_path):
            os.makedirs(folder_path)
            print(f"  已创建文件夹: {folder_name}")
        else:
            print(f"  已存在，跳过文件夹: {folder_name}")

    # 2. 创建 case_meta.json
    meta_path = os.path.join(case_path, "case_meta.json")
    if not os.path.exists(meta_path):
        meta_data = default_case_meta.copy()
        meta_data["case_id"] = name  # 自动写入当前 case 文件夹名

        with open(meta_path, "w", encoding="utf-8") as f:
            json.dump(meta_data, f, ensure_ascii=False, indent=4)

        print("  已创建文件: case_meta.json")
    else:
        print("  已存在，跳过文件: case_meta.json")

print("\n全部处理完成。")
```
