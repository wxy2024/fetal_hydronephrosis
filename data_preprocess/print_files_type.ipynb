```python
##展示一个文件夹下所有子文件的文件格式

from pathlib import Path
from collections import Counter

# =========================
# 1. 修改为你的根目录
# =========================
ROOT_DIR = Path(r"G:\wxy_codes\Fetal_hydronephrosis\1第一批数据\batch1data_3")

# =========================
# 2. 递归扫描所有文件
# =========================
suffix_counter = Counter()

for file_path in ROOT_DIR.rglob("*"):
    if file_path.is_file():
        suffix = file_path.suffix.lower()
        
        # 没有后缀的文件
        if suffix == "":
            suffix = "[无后缀]"
        
        suffix_counter[suffix] += 1

# =========================
# 3. 输出结果
# =========================
print(f"共发现 {len(suffix_counter)} 种文件格式：\n")

for suffix, count in sorted(suffix_counter.items()):
    print(f"{suffix}：{count} 个")
```

```python
##打印出所有的jpg文件的地址
from pathlib import Path

root = Path(r"G:\wxy_codes\Fetal_hydronephrosis\1第一批数据\batch1data_3")

jpg_files = []

for file_path in root.rglob("*"):
    if file_path.is_file() and file_path.suffix.lower() == ".jpg":
        jpg_files.append(file_path)

print(f"共找到 {len(jpg_files)} 个 jpg 文件：\n")

for jpg_path in jpg_files:
    print(jpg_path)
```
